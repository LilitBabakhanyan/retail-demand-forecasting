## Dataset Overview
The Walmart Store Sales Forecasting dataset captures weekly sales data for 45 Walmart stores across different regions, broken down by individual departments. The dataset shows the relationship between sales performance and various external factors, providing a comprehensive view of retail operations over a nearly three-year period (February 2010 to November 2012).

**What the dataset shows:**

*   Sales patterns over time for each store-department combination, including seasonal trends and spikes during holiday weeks
*   The impact of promotional markdowns on consumer purchasing behavior, particularly surrounding major holidays like Super Bowl, Thanksgiving, and Christmas
* How external economic factors (CPI, unemployment rate) and regional conditions (temperature, fuel prices) correlate with retail sales

**What the dataset contains:**

* Historical sales records with fields for Store, Department, Date, Weekly_Sales, and a holiday indicator

* Store metadata including anonymized store type and size

* External features such as Temperature, Fuel_Price, Unemployment rate, CPI, and five anonymized markdown fields (MarkDown1-5)

* Holiday identification for weeks containing major events, which are weighted five times higher in evaluation

## Train Dataset Columns
| Column | Meaning | Data Type | Role |
|---|---|---|---|
| Store | Store id | Integer | Grouping variable |
| Dept | Department id | Integer | Grouping variable |
| Date | Week of observation | Object (date) | Time index |
| Weekly_Sales | Weekly revenue | Float | Target variable |
| IsHoliday | Holiday indicator | Boolean | Feature |


## Features Dataset Columns

| Column | Meaning | Data Type | Role |
|---|---|---|---|
| Store | Store id | Integer | Grouping variable |
| Date | Week of observation | Object (date) | Time index |
| Temperature | Average weekly temperature | Float | Feature |
| Fuel_Price | Regional fuel price | Float | Feature |
| MarkDown1 | Promotional markdown amount 1 | Float | Feature |
| MarkDown2 | Promotional markdown amount 2 | Float | Feature |
| MarkDown3 | Promotional markdown amount 3 | Float | Feature |
| MarkDown4 | Promotional markdown amount 4 | Float | Feature |
| MarkDown5 | Promotional markdown amount 5 | Float | Feature |
| CPI | Consumer Price Index | Float | Economic feature |
| Unemployment | Regional unemployment rate | Float | Economic feature |
| IsHoliday | Holiday indicator | Boolean | Feature |

- **MarkDown1–5:** Promotional markdown amounts applied during that week.  
  Example: MarkDown1 = 10382.9 indicates a total discount of $10,382.9 applied for that promotion type. NaN means no promotion was applied that week.
